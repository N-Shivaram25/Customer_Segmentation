# 1. Create preprocessing copy
Keep the original EDA dataframe unchanged.

```python

df_prep = df.copy()
```

#  2. Remove irrelevant/constant variables
 ID identifies a customer but does not describe customer behavior.
 Z_CostContact and Z_Revenue are constant, so they provide no information
 for distinguishing customers.

```python

columns_to_remove = ["ID", "Z_CostContact", "Z_Revenue"]
existing_columns_to_remove = [
    column for column in columns_to_remove if column in df_prep.columns
]
df_prep = df_prep.drop(columns=existing_columns_to_remove)

print("Removed columns:", existing_columns_to_remove)

```


#  3. Missing-value treatment
Inspect missing values before deciding how to handle them.

```python

missing_before = df_prep.isna().sum()
print("Missing-value counts before treatment:")
display(missing_before[missing_before > 0])

columns_with_missing = missing_before[missing_before > 0].index.tolist()
print("Columns containing missing values:", columns_with_missing)
```

Income is a numerical candidate feature. Use its median only when it has

missing values; do not replace missing values with zero.

```python

if "Income" in df_prep.columns and df_prep["Income"].isna().any():
    income_median = df_prep["Income"].median()
    df_prep["Income"] = df_prep["Income"].fillna(income_median)
    print(f"Imputed missing Income values with the median: {income_median}")

print("Missing-value counts after Income treatment:")
display(df_prep.isna().sum()[df_prep.isna().sum() > 0])
```


#  4. Date and tenure handling
 A raw date is not a suitable direct numerical input to K-Means.
 
 Preserve the existing Tenure_Days feature if the notebook already created it.

```python


if "Tenure_Days" not in df_prep.columns and "Dt_Customer" in df_prep.columns:
    df_prep["Dt_Customer"] = pd.to_datetime(
        df_prep["Dt_Customer"], errors="coerce"
    )
    reference_date = df_prep["Dt_Customer"].max()

    if pd.notna(reference_date):
        df_prep["Tenure_Days"] = (
            reference_date - df_prep["Dt_Customer"]
        ).dt.days
    else:
        print("Tenure_Days was not created: no valid Dt_Customer dates were found.")

```

 Drop the raw date from the preprocessing dataframe; keep Tenure_Days if present.


```python
if "Dt_Customer" in df_prep.columns:
    df_prep = df_prep.drop(columns=["Dt_Customer"])
```


Year_Birth has potentially anomalous values, so inspect it but do not use it

directly or automatically remove any rows.

```python

if "Year_Birth" in df_prep.columns:
    print("Year_Birth summary:")
    display(df_prep["Year_Birth"].describe())
    print("Earliest Year_Birth values for inspection:")
    display(df_prep["Year_Birth"].sort_values().head(10))
    print("Year_Birth values below 1900 for inspection:")
    display(df_prep.loc[df_prep["Year_Birth"] < 1900, ["Year_Birth"]].head(20))
```


 The existing EDA created Age using the current year. Flag it for review;

do not use this potentially inaccurate Age feature for the first model.

```python

if "Age" in df_prep.columns:
    print(
        "Age exists and is flagged for review because the existing EDA "
        "calculated it using the current year. It will not be selected."
    )

```

#  5. Candidate clustering features

 Select only candidate features that exist. Total_Spending summarizes product

 spending, and Total_Purchases summarizes channel purchases. To avoid

 redundancy, this candidate set uses each aggregate instead of also adding
 
 all of its component columns.

```python


candidate_features = [
    "Income",
    "Recency",
    "Total_Spending",
    "Total_Purchases",
    "NumDealsPurchases",
    "NumWebVisitsMonth",
    "Total_Children",
]
features = [column for column in candidate_features if column in df_prep.columns]
missing_candidates = [
    column for column in candidate_features if column not in df_prep.columns
]

print("Candidate features not found:", missing_candidates)
print("Selected candidate features:", features)

feature_reasons = {
    "Income": "Household income provides a customer spending-capacity measure.",
    "Recency": "Days since the last purchase describes purchase activity.",
    "Total_Spending": "Summarizes spending across product categories.",
    "Total_Purchases": "Summarizes purchase frequency across channels.",
    "NumDealsPurchases": "Describes use of discount purchases.",
    "NumWebVisitsMonth": "Describes recent website engagement.",
    "Total_Children": "Describes household composition.",
}
for feature in features:
    print(f"- {feature}: {feature_reasons[feature]}")

```


#  6. Data validation

 Build X from the selected candidate features only.

```python

X = df_prep[features].copy()

print("X shape:", X.shape)
print("X data types:")
display(X.dtypes)
print("Missing values in X:")
display(X.isna().sum())
print("Descriptive statistics for X:")
display(X.describe().T)
print("Duplicate rows in X:", X.duplicated().sum())

```


#  7. Outlier analysis

 Boxplots help identify potential extreme values; they do not establish that

 those values are errors. Genuine high-value customers should not be removed

 automatically. Investigate data errors separately from genuine behavior.

```python

if features:
    plt.figure(figsize=(max(10, len(features) * 1.5), 6))
    sns.boxplot(data=X)
    plt.title("Potential Outliers in Candidate Clustering Features")
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()

print(
    "If investigation shows that treatment is necessary, consider a defensible "
    "method such as winsorization/capping. Compare and document its impact "
    "before applying it."
)
```



#  8. Feature scaling

 StandardScaler requires numeric features without missing values. Do not

silently fill any remaining missing candidate values.

```python

scaling_completed = False

if not features:
    print("Scaling not completed: no candidate features were found.")
elif X.isna().any().any():
    print("Scaling not completed: selected features still contain missing values.")
    display(X.isna().sum()[X.isna().sum() > 0])
elif not all(pd.api.types.is_numeric_dtype(dtype) for dtype in X.dtypes):
    print("Scaling not completed: one or more selected features are not numeric.")
else:
    from sklearn.preprocessing import StandardScaler

    scaler = StandardScaler()
    X_scaled_values = scaler.fit_transform(X)
    X_scaled = pd.DataFrame(
        X_scaled_values,
        columns=X.columns,
        index=X.index,
    )
    scaling_completed = True

    print("Scaled feature means (approximately 0):")
    display(X_scaled.mean())
    print("Scaled feature standard deviations (population; approximately 1):")
    display(X_scaled.std(ddof=0))


```

#  9. Preprocessing summary


```python
print(f"Number of customers: {len(df_prep)}")
print(f"Number of candidate clustering features: {len(features)}")
print("Features selected:", features)
print(f"Missing values remaining in df_prep: {int(df_prep.isna().sum().sum())}")
print(f"Missing values remaining in X: {int(X.isna().sum().sum())}")
print(f"Scaling completed: {scaling_completed}")

```
